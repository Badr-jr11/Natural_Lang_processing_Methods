```python
import spacy
from nltk.stem import PorterStemmer

nlp = spacy.load("en_core_web_sm")
stemmer = PorterStemmer()

def preprocess(text):
      doc = nlp(text.lower())
      tokens = [
      t for t in doc
      if not t.is_punct
      and not t.is_stop
      and not t.is_space
      ]
      lemmas = [t.lemma_ for t in tokens]
      stems = [stemmer.stem(t.text) for t in tokens]
      return lemmas, stems

text = "the embedded atom model, embedded-atom method"
lemmas, stems = preprocess(text)

print("Lemmas:", lemmas)
print("Stems: ", stems)
```
